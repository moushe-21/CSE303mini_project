```python
import pandas as pd

df = pd.read_csv("laptop_price - dataset.csv")
```

```python
# Check column names to confirm
print(df.columns)
```

```python
# Define continuous columns exactly as in dataset
continuous_cols = ["Inches", "CPU_Frequency (GHz)", "RAM (GB)", "Weight (kg)", "Price (Euro)"]
```

```python
print(df[continuous_cols].agg(["mean", "median", "var", "std"]))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
df["Company"].value_counts().plot(kind="bar")
plt.title("Distribution of Laptop Brands")
plt.xlabel("Company")
plt.ylabel("Count")
plt.show()
```

```python
type_counts = df['TypeName'].value_counts()
plt.figure(figsize=(5,5))
plt.pie(type_counts, labels=type_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Distribution of Laptop Types")
plt.show()

plt.figure(figsize=(8,6))
sns.heatmap(df[continuous_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Continuous Features")
plt.show()
```
